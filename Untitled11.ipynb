Word Frequency Analysis

```python
import re
from collections import Counter
from nltk.corpus import stopwords
STOP = set(stopwords.words("english"))

def tokenize(t):
    t = t.lower()
    return [w for w in re.findall(r"[a-z']+", t) if w not in STOP and len(w)>1]

q_counter = Counter()
a_counter = Counter()

if question_col:
    for q in df[question_col].dropna().astype(str):
        q_counter.update(tokenize(q))

if answer_col:
    for a in df[answer_col].dropna().astype(str):
        a_counter.update(tokenize(a))

print("Top Question Words:")
q_counter.most_common(20)
```

```python
print("Top Answer Words:")
a_counter.most_common(20)
```

```python
def plot_top(counter, title):
    top = counter.most_common(20)
    if len(top)==0:
        return
    words, counts = zip(*top)
    plt.figure(figsize=(10,5))
    sns.barplot(x=list(counts), y=list(words))
    plt.title(title)
    plt.show()

plot_top(q_counter, "Top Question Words")
plot_top(a_counter, "Top Answer Words")
```

```python
# Diagnostic + extractor
import io
from PIL import Image
import numpy as np
from tqdm.notebook import tqdm

def to_pil(img_field):
    # handle PIL.Image
    if hasattr(img_field, "size") and hasattr(img_field, "mode") and isinstance(img_field, Image.Image):
        return img_field.convert("RGB")
    # numpy array
    if isinstance(img_field, np.ndarray):
        arr = img_field
        if arr.dtype != np.uint8:
            try:
                arr = ((arr - arr.min()) / (arr.max() - arr.min() + 1e-9) * 255).astype(np.uint8)
            except Exception:
                arr = arr.astype(np.uint8)
        if arr.ndim == 2:
            arr = np.stack([arr]*3, axis=-1)
        return Image.fromarray(arr).convert("RGB")
    # bytes
    if isinstance(img_field, (bytes, bytearray)):
        return Image.open(io.BytesIO(img_field)).convert("RGB")
    # dict-like (HF datasets Image often)
    if isinstance(img_field, dict):
        # typical keys: 'path', 'bytes', 'array'
        if 'bytes' in img_field and img_field['bytes'] is not None:
            return Image.open(io.BytesIO(img_field['bytes'])).convert("RGB")
        if 'path' in img_field and img_field['path'] is not None:
            p = img_field['path']
            try:
                return Image.open(p).convert("RGB")
            except Exception as e:
                # path might be bytes; try bytes fallback
                if isinstance(p, (bytes, bytearray)):
                    return Image.open(io.BytesIO(p)).convert("RGB")
                # otherwise re-raise later
        if 'array' in img_field:
            return to_pil(np.array(img_field['array']))
    # datasets Image object or wrapper with .to_pil / .to_pillow / .to_numpy
    if hasattr(img_field, "to_pil"):
        try:
            return img_field.to_pil().convert("RGB")
        except Exception:
            pass
    if hasattr(img_field, "to_pillow"):
        try:
            return img_field.to_pillow().convert("RGB")
        except Exception:
            pass
    if hasattr(img_field, "to_numpy"):
        try:
            return to_pil(img_field.to_numpy())
        except Exception:
            pass
    # fallback: attempt PIL open on str path
    if isinstance(img_field, str):
        try:
            return Image.open(img_field).convert("RGB")
        except Exception:
            pass
    raise ValueError(f"Unsupported image field type: {type(img_field)}")

# Now try to extract sizes for the whole dataframe (safe: catch exceptions and report them)
sizes = []
errors = 0
max_iter = len(df)  # change to limit if very large
for i, imgv in enumerate(tqdm(df["image"].iloc[:max_iter], desc="extracting sizes")):
    try:
        pil = to_pil(imgv)
        sizes.append((pil.width, pil.height))
    except Exception as e:
        errors += 1
        # optionally print first error to debug
        if errors <= 5:
            print(f"[error sample {errors}] idx {i}: {e}")
        continue

print("Total rows inspected:", max_iter)
print("Successfully extracted sizes:", len(sizes))
print("Errors:", errors)
```

```python
# Robust image display (replacement for Cell 10)
import random, io
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

def display_image_field(img_field):
    """
    Accepts many possible image representations:
      - PIL.Image.Image
      - numpy array (H x W) or (H x W x C)
      - bytes (encoded image)
      - dict from HF datasets image (may contain 'path' or 'bytes')
      - datasets.Image object that behaves like PIL or dict
    Returns a PIL Image
    """
    # 1) PIL image already
    if isinstance(img_field, Image.Image):
        return img_field

    # 2) numpy array
    if isinstance(img_field, np.ndarray):
        # ensure numeric dtype suitable for imshow
        if img_field.dtype != np.uint8:
            # try to normalize to 0-255
            try:
                arr = img_field.astype(float)
                arr = 255 * (arr - arr.min()) / (arr.max() - arr.min()) if arr.max()!=arr.min() else arr
                arr = np.clip(arr, 0, 255).astype(np.uint8)
            except Exception:
                arr = img_field.astype(np.uint8)
        else:
            arr = img_field
        # if grayscale (H,W) convert to RGB
        if arr.ndim == 2:
            arr = np.stack([arr]*3, axis=-1)
        return Image.fromarray(arr)

    # 3) bytes (encoded image like PNG/JPEG)
    if isinstance(img_field, (bytes, bytearray)):
        try:
            return Image.open(io.BytesIO(img_field)).convert("RGB")
        except Exception as e:
            raise RuntimeError("Cannot open image bytes: " + str(e))

    # 4) dict-like object (HF datasets Image often serialized as dict with 'path' or 'bytes')
    if isinstance(img_field, dict):
        # common keys: 'path', 'bytes', 'array'
        if 'bytes' in img_field and img_field['bytes'] is not None:
            return Image.open(io.BytesIO(img_field['bytes'])).convert("RGB")
        if 'path' in img_field and img_field['path'] is not None:
            # path might be local file path or dataset store path
            try:
                return Image.open(img_field['path']).convert("RGB")
            except Exception as e:
                # fallback: try to open as bytes if value is bytes-like
                try:
                    val = img_field['path']
                    if isinstance(val, (bytes, bytearray)):
                        return Image.open(io.BytesIO(val)).convert("RGB")
                except Exception:
                    pass
        # some HF images store a nested 'data' or 'array'
        if 'array' in img_field:
            return display_image_field(np.array(img_field['array']))
        # if nothing useful
        raise RuntimeError("Unrecognized dict image structure; keys: " + ", ".join(img_field.keys()))

    # 5) datasets.Image or other wrapper that may be convertible to PIL via .to_pil or .convert
    # try common duck-typed approaches:
    try:
        # e.g., datasets.Image objects may expose .to_pil or .convert
        if hasattr(img_field, "to_pil"):
            pil = img_field.to_pil()
            if isinstance(pil, Image.Image):
                return pil
        if hasattr(img_field, "convert"):
            pil = img_field.convert("RGB")
            if isinstance(pil, Image.Image):
                return pil
        # try turning into numpy
        if hasattr(img_field, "to_numpy"):
            arr = img_field.to_numpy()
            return display_image_field(arr)
    except Exception:
        pass

    raise RuntimeError("Unsupported image field type: " + str(type(img_field)))


# Choose a random sample and display
idx = random.randint(0, len(df)-1)
sample = df.iloc[idx]

print("Question ID:", sample.get("question_id", "N/A"))
print("Question:", sample.get("question", "N/A"))

# 'answer' might be a list/ndarray of captions or a single string
answers = sample.get("answer", None)
if isinstance(answers, (list, tuple, np.ndarray)):
    # show a few
    print("Answer (examples):")
    for i, a in enumerate(answers[:10]):
        print(f"  {i+1}. {a}")
else:
    print("Answer:", answers)

# attempt to display the image
try:
    pil_img = display_image_field(sample["image"])
    plt.figure(figsize=(6,6))
    plt.imshow(pil_img)
    plt.axis("off")
    plt.title(f"Sample index: {idx}")
    plt.show()
except Exception as e:
    print("Could not display image:", repr(e))
    # print the image field type/keys to help debug
    imgf = sample.get("image", None)
    print("image field type:", type(imgf))
    if isinstance(imgf, dict):
        print("image dict keys:", list(imgf.keys()))
```
